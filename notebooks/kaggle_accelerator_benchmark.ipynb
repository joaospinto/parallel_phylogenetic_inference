# Parallel phylogenetic tree inference: accelerator validation and benchmark

Start a fresh Kaggle notebook with an NVIDIA or AMD GPU and Internet access. Until the repositories are public, run `scripts/package_notebook_sources.sh`, upload the resulting ZIP as a Kaggle dataset, and attach it to this notebook. The launcher detects the GPU, compiles both Linux accelerator backends from pinned SDKs, and executes the matching CUDA or ROCm implementation in FP64 and FP32. It validates likelihood, marginal, MAP, and posterior-sampling inference before timing synthetic scaling cases, all 325 PANDIT 17.0 families with at least 100 tips, and the public 11,638-taxon Fish Tree of Life alignment. The conventional CPU implementation is measured in the same precision; BEAGLE CPU is included on both vendors and BEAGLE CUDA on NVIDIA. Synthetic rows use 15 repetitions and empirical rows use three. JIT compilation is not involved. Site batches increase geometrically up to the complete alignment or each method's capacity. The code cell may override `TREE_HMM_BENCHMARK_SECTIONS` with any subset of `validation synthetic fish pandit`, together with the documented precision, repetition, sanitizer, and `TREE_HMM_SKIP_*` controls. Cache reuse is keyed by all packaged source revisions.

```python
from pathlib import Path
import subprocess
import zipfile

inputs = Path("/kaggle/input")
bundle = next(inputs.rglob("parallel_tree_inference_sources.zip"), None)
launcher = next(
    inputs.rglob("parallel_phylogenetic_inference/scripts/kaggle_accelerator_notebook.sh"),
    None,
)

if bundle is not None:
    with zipfile.ZipFile(bundle) as archive:
        script = archive.read(
            "parallel_phylogenetic_inference/scripts/kaggle_accelerator_notebook.sh"
        )
    subprocess.run(
        ["bash", "-s", "--", str(bundle)], input=script, check=True
    )
elif launcher is not None:
    source_root = launcher.parents[2]
    subprocess.run(["bash", str(launcher), str(source_root)], check=True)
else:
    raise FileNotFoundError(
        "Attach the parallel_tree_inference_sources input before running"
    )
```

The complete machine description, validation log, capacity boundaries, and benchmark CSV rows are checkpointed in `/kaggle/working/parallel_phylogenetics_cuda_report.txt` or `/kaggle/working/parallel_phylogenetics_rocm_report.txt`, according to the detected GPU. Download that file even if a runtime interruption occurs; a later source bundle can embed it and resume only source-compatible missing work.
